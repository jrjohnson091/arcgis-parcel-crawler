```python
square = {
    "rings": [[
        [0, 0],
        [100, 0],
        [100, 100],
        [0, 100],
        [0, 0],
    ]]
}
```

```python
import matplotlib.pyplot as plt

ring = square["rings"][0]

x = [p[0] for p in ring]
y = [p[1] for p in ring]

plt.figure(figsize=(5, 5))
plt.plot(x, y)
plt.axis("equal")
plt.show()
```
